```python
# ============================================================
# SHAP STEP 2: EXACT FEATURE ORDER VERIFICATION
# ============================================================

import json
import duckdb

# 1. Inspect expected structure
print("EXPECTED STRUCTURE:")
print(json.dumps(
    params.get("expected_structure", {}),
    indent=2
))

# 2. Get actual transformed training columns
con = duckdb.connect()

train_schema = con.execute(
    f"DESCRIBE SELECT * FROM read_parquet('{TRAIN_FILE}')"
).fetchdf()

validation_schema = con.execute(
    f"DESCRIBE SELECT * FROM read_parquet('{VALIDATION_FILE}')"
).fetchdf()

train_columns = train_schema["column_name"].tolist()
validation_columns = validation_schema["column_name"].tolist()

# 3. Identify core and predictor columns
CORE_COLUMNS = [
    "loan_id",
    "reporting_date",
    "target_6m"
]

predictor_columns = [
    col for col in train_columns
    if col not in CORE_COLUMNS
]

# 4. Verify exact schema order
print("\nSCHEMA AUDIT")
print("Train total columns:", len(train_columns))
print("Validation total columns:", len(validation_columns))
print("Predictor count:", len(predictor_columns))
print("HGB expected features:", hgb_model.n_features_in_)

print(
    "Train/validation exact column order:",
    train_columns == validation_columns
)

print(
    "Model feature count matches:",
    len(predictor_columns) == hgb_model.n_features_in_
)

print("\nCORE COLUMNS:", CORE_COLUMNS)

print("\nORDERED PREDICTOR FEATURES:")
for i, col in enumerate(predictor_columns, start=1):
    print(f"{i:03d}. {col}")

con.close()
```

```python
# ============================================================
# SHAP STEP 3: MODEL TRAINING FEATURE-ORDER AUDIT
# ============================================================

import json
from pathlib import Path

# A. Check whether the saved model retained feature names
print("MODEL FEATURE METADATA")

print(
    "Has feature_names_in_:",
    hasattr(hgb_model, "feature_names_in_")
)

if hasattr(hgb_model, "feature_names_in_"):
    print("Saved feature names:")
    print(list(hgb_model.feature_names_in_))

# B. Inspect actual transformed predictor columns
print("\nACTUAL FEATURE COUNT:", len(predictor_columns))

print("\nDATE-RELATED FEATURES:")
for col in predictor_columns:
    if "reporting_year" in col or "reporting_month" in col:
        print(col)

# C. Inspect the original training notebook
NOTEBOOK_DIR = PROJECT_DIR / "notebooks"

print("\nMODEL TRAINING CODE SEARCH")

matches = []

for nb_path in NOTEBOOK_DIR.rglob("*.ipynb"):
    try:
        with open(nb_path, "r", encoding="utf-8") as f:
            notebook = json.load(f)

        for cell_no, cell in enumerate(notebook.get("cells", [])):
            if cell.get("cell_type") != "code":
                continue

            source = "".join(cell.get("source", []))

            keywords = [
                "HistGradientBoostingClassifier",
                "hgb_model.fit",
                "model.fit",
                "X_train",
                "train_transformed"
            ]

            if any(word in source for word in keywords):
                matches.append(
                    (nb_path.name, cell_no, source)
                )

    except Exception as e:
        print("Could not inspect:", nb_path.name, str(e))

for notebook_name, cell_no, source in matches:
    print("\n" + "=" * 70)
    print("NOTEBOOK:", notebook_name, "| CELL:", cell_no)
    print(source[:5000])

if not matches:
    print("No matching model-training cells found.")
```

```python
import json
from pathlib import Path

NOTEBOOK_DIR = PROJECT_DIR / "notebooks"

keywords = [
    "HistGradientBoostingClassifier",
    "hgb_model.fit",
    "X_train =",
    "feature_columns",
    "predictor_columns",
    "training_features",
    "reporting_year_1",
    "reporting_month_1"
]

for nb_path in NOTEBOOK_DIR.rglob("*.ipynb"):

    with open(nb_path, "r", encoding="utf-8") as f:
        notebook = json.load(f)

    for cell_no, cell in enumerate(notebook.get("cells", [])):

        if cell.get("cell_type") != "code":
            continue

        lines = "".join(cell.get("source", [])).splitlines()

        matched = [
            i for i, line in enumerate(lines)
            if any(k.lower() in line.lower() for k in keywords)
        ]

        if matched:
            print("\n" + "=" * 65)
            print("NOTEBOOK:", nb_path.name)
            print("CELL:", cell_no)

            shown = set()

            for i in matched:
                for j in range(max(0, i-2), min(len(lines), i+3)):
                    if j not in shown:
                        print(f"{j+1}: {lines[j]}")
                        shown.add(j)
```

```python
# ============================================================
# FINAL HGB FEATURE-ORDER VERIFICATION
# ============================================================

print("MODEL HAS FEATURE NAMES:",
      hasattr(hgb_model, "feature_names_in_"))

if hasattr(hgb_model, "feature_names_in_"):

    model_features = list(hgb_model.feature_names_in_)

    print("MODEL FEATURE COUNT:", len(model_features))
    print("DATASET FEATURE COUNT:", len(predictor_columns))

    print(
        "EXACT FEATURE ORDER MATCH:",
        model_features == predictor_columns
    )

    if model_features != predictor_columns:
        print("\nMISMATCH DETAILS:")

        for i, (a, b) in enumerate(
            zip(model_features, predictor_columns)
        ):
            if a != b:
                print("First mismatch at position:", i + 1)
                print("Model:", a)
                print("Dataset:", b)
                break

else:
    print("Feature names were not retained by the model.")
```

```python
from pathlib import Path
import os

PROJECT_DIR = Path(
    r"C:\Users\madhu\OneDrive\Desktop\MTP - Dynamic Credit Risk Early Warning"
)

MODELS_DIR = PROJECT_DIR / "models"

print("Project exists:", PROJECT_DIR.exists())
print("Models folder exists:", MODELS_DIR.exists())

for file in sorted(MODELS_DIR.iterdir()):
    if file.is_file():
        print(
            file.name,
            round(file.stat().st_size / 1024, 2),
            "KB"
        )
```

```python
import joblib

HGB_PATH = MODELS_DIR / "hist_gradient_boosting_baseline.joblib"

hgb_model = joblib.load(HGB_PATH)

print("HGB MODEL LOADED SUCCESSFULLY")
print("Model type:", type(hgb_model))
print("Number of iterations:", hgb_model.n_iter_)
```

```python
import json
import pandas as pd

print("CSV FILE CHECK")

for file in MODELS_DIR.glob("*.csv"):
    try:
        df = pd.read_csv(file, nrows=5)
        print("PASS:", file.name, "| Columns:", len(df.columns))
    except Exception as e:
        print("FAIL:", file.name, "|", str(e))

print("\nJSON FILE CHECK")

for file in MODELS_DIR.glob("*.json"):
    try:
        with open(file, "r", encoding="utf-8") as f:
            json.load(f)
        print("PASS:", file.name)
    except Exception as e:
        print("FAIL:", file.name, "|", str(e))
```

```python
import joblib
import json

# Load Logistic Regression
LR_PATH = MODELS_DIR / "baseline_logistic_sgd.joblib"
lr_model = joblib.load(LR_PATH)

print("LOGISTIC REGRESSION: PASS")
print("Model type:", type(lr_model))
print("Number of input features:", lr_model.n_features_in_)

# HGB model information
print("\nHGB MODEL INFORMATION")
print("Number of input features:", hgb_model.n_features_in_)
print("Number of iterations:", hgb_model.n_iter_)
print("Classes:", hgb_model.classes_)

# Inspect model registry
print("\nMODEL REGISTRY")
with open(MODELS_DIR / "model_registry.json", "r") as f:
    registry = json.load(f)

print(json.dumps(registry, indent=2))
```
